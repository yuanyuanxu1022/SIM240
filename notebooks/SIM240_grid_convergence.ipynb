```python
import pandas as pd

rows = []

for grid in ["dx5", "dx2p5", "dx1p25"]:
    for case_folder in ["Smooth75", "EX240_parallel", "EX240_perpendicular"]:

        f = (
            latest
            / "runs"
            / grid
            / case_folder
            / "steady"
            / "result_summary.csv"
        )

        if f.exists():
            df = pd.read_csv(f)

            # 只补充目录标签，不改原始物理数据
            df["grid"] = grid
            df["case_folder"] = case_folder

            rows.append(df)

summary = pd.concat(rows, ignore_index=True)

display(
    summary[
        [
            "grid",
            "case_folder",
            "Case",
            "direction",
            "dx",
            "J",
            "Keff",
            "R_eff",
            "Mach",
            "mass_error",
        ]
    ]
)
```
